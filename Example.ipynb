## Compared rail and road density between countries

### First, we can look at the distribution of the road density variable for each country.

```python
print(summarize_distribution("Zambia", "road"))
print(summarize_distribution("Zimbabwe", "road"))
print(summarize_distribution("Ghana", "road"))
```

### We can also plot road density for all three countries to compare. Figures are saved to an output folder.

```python
plot_density(country="Zambia", var="road")
plot_density(country="Zimbabwe", var="road")
plot_density(country="Ghana", var="road")
```

### What about rail density?

```python
plot_density(country="Zambia", var="rail")
plot_density(country="Zimbabwe", var="rail")
plot_density(country="Ghana", var="rail")
```

### From the plots, we see that rail and road density differ significantly within countries. How can we compare?
